# 1. What is Unsupervised Learning?

### Concept
In machine learning, tasks are broadly categorized based on the data they use:
* **Supervised Learning**: The algorithm learns from **labeled data**. It predicts known targets (e.g., house price).
* **Unsupervised Learning**: The algorithm learns from **unlabeled data**. It attempts to find hidden structures without known targets.

### Why Clustering is Useful
Clustering is the primary unsupervised learning technique. The objective is:
**Input data → Find hidden groups → Assign similar points to clusters**

> Key Idea: While supervised learning matches inputs to a predetermined outcome, clustering discovers the natural groupings within the data. We then interpret what those groups mathematically represent.


# 2. What is K-Means Clustering?

### Concept
* **What is clustering?** Grouping items such that items in the same group are more similar to each other.
* **What is K-Means?** A centroid-based clustering algorithm dividing data into a specific number of non-overlapping clusters.
* **Why it is called K-Means?** $K$ refers to the number of clusters you want. "Means" refers to calculating the average (the geometric center) of the data points to form the cluster center.

### Real-world applications
* **Customer segmentation**: Grouping customers by similar purchasing patterns.
* **Market segmentation**: Identifying diverse demographics for targeted marketing.
* **Grouping products**: Finding similarities between items in an e-commerce catalog.
* **Image color compression**: Reducing colors in an image down to the most standard palettes.
* **Document grouping**: Automatically tagging text topics.

### Basic Intuition
Imagine mapping data as points on a piece of paper. If you visually see distinct "clouds" of dots, each cloud is a cluster. K-Means mathematically isolates these clouds.


# 3. What does K Mean?

### Concept
**$K = 	ext{number of clusters}$**

The standard K-Means algorithm requires you to define $K$ before it starts training.

* **K = 2** → The algorithm will force the data into exactly two clusters.
* **K = 3** → The algorithm will find three clusters.
* **K = 5** → The algorithm will find five clusters.

> Important: Choosing K is critical. The wrong K can unnaturally force distinct groups together, or artificially split a uniform group into pieces.


# 4. K-Means Algorithm

### How it works step-by-step
The K-Means algorithm follows an iterative process:

1. **Choose K**: Define the number of clusters to form.
2. **Initialize centroids**: Randomly drop $K$ points (centroids) into the dataset.
3. **Assign each data point to the nearest centroid**: Measure the distance between every data point and every centroid. Assign points to the closest centroid.
4. **Recalculate centroids**: Now that points are grouped, calculate the mean position of all points inside each cluster, and move the centroid to that mean.
5. **Repeat assignment and update**: Because centroids moved, distances changed! Reassign points to the *new* closest centroid, and calculate the new mean again.
6. **Stop**: Halt when the centroids mathematically stabilize (they stop moving) or maximum iterations are hit.

> Key Idea: K-Means is a loop of Assign -> Update -> Assign -> Update until stability.


# 5. Centroids

### Concept
* **What is a centroid?** The "mean" or center of a cluster. It's the geometric representation of the group.
* **How centroid is calculated**: You add up all coordinates along each dimension and divide by the total number of points in that cluster. Over many dimensions, it's the element-wise average.
* **Why centroid represents a cluster**: It acts as the "prototype" to which all other points in that cluster are naturally closest.


```python
# Example: Calculating a centroid manually using NumPy
# A simple 2D dataset with 4 points
points = np.array([
    [2, 3],
    [5, 4],
    [3, 6],
    [4, 2]
])

# Mean across the columns (axis=0)
centroid = np.mean(points, axis=0)

print("Data Points:\n", points)
print("\nCalculated Centroid:", centroid)
```

# 6. Distance in K-Means

### Euclidean Distance
K-Means natively measures proximity using basic straight-line mathematical distance. In 2D, the formula is:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

### Why Distance Determines Assignment
The distance determines similarity. A data point is grouped with whichever centroid it is closest to mathematically.


```python
# Example: Determining the nearest centroid using Distance
point = np.array([5, 5])

centroid_A = np.array([2, 2])
centroid_B = np.array([7, 7])

# Calculate Euclidean distance manually
dist_A = np.sqrt(np.sum((point - centroid_A)**2))
dist_B = np.sqrt(np.sum((point - centroid_B)**2))

print(f"Distance to Centroid A: {dist_A:.2f}")
print(f"Distance to Centroid B: {dist_B:.2f}")

if dist_A < dist_B:
    print("\nResult: Assign point to Cluster A")
else:
    print("\nResult: Assign point to Cluster B")
```

# 7. K-Means from Scratch

### Concept
To visualize the iterations clearly, let's write a simple implementation of K-Means entirely from scratch using only NumPy.


```python
# Implement simple K-Means from scratch
# 1. Dataset
np.random.seed(42)
group1 = np.random.randn(50, 2) + np.array([1, 1])
group2 = np.random.randn(50, 2) + np.array([6, 6])
X = np.vstack([group1, group2])

# 2. Initialize K
K = 2

# 3. Initialize centroids randomly by picking K random points from our data
random_indices = np.random.choice(X.shape[0], K, replace=False)
centroids = X[random_indices]

labels = np.zeros(X.shape[0])
max_iters = 10

print(f"Initial Centroids:\n{centroids}")

for i in range(max_iters):
    # 4. Calculate distances and Assign clusters
    for j, point in enumerate(X):
        distances = [np.linalg.norm(point - c) for c in centroids]
        labels[j] = np.argmin(distances) # index of min distance
        
    old_centroids = centroids.copy()
    
    # 5. Update centroids
    for k in range(K):
        # Filter points assigned to cluster k
        cluster_points = X[labels == k]
        if len(cluster_points) > 0:
            centroids[k] = np.mean(cluster_points, axis=0)
            
    # 6. Stop when stable
    if np.all(old_centroids == centroids):
        print(f"\nConverged stably at iteration {i}!")
        break

# Visualize clusters after training
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', label='Assigned Data')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, marker='X', edgecolor='k', label='Final Centroids')
plt.title("K-Means from Scratch Result")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```

# 8. K-Means with Scikit-learn

### Concept
In a real-world scenario, you will always use the optimized version inside `scikit-learn`. Let's demonstrate modern library usage.


```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Create dataset
X_sk, y_sk = make_blobs(n_samples=300, centers=4, cluster_std=0.8, random_state=42)

# 2. Inspect data
print("Data shape:", X_sk.shape)
print("First 3 points:\n", X_sk[:3])

# 3. Select features (We use the entirety of X_sk which has 2 features)

# 4. Create KMeans model
model = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

# 5. Fit model
model.fit(X_sk)

# 6. Get labels for every data point
labels_sk = model.labels_

# 7. Get calculated cluster centers
centers_sk = model.cluster_centers_

print("\nFirst 10 Labels assigned:", labels_sk[:10])
print("\nCalculated Centroids:\n", centers_sk)
```

### Important Scikit-Learn Parameters
* `n_clusters`: Matches our $K$ value.
* `random_state`: Makes the internal random initialization completely reproducible across different runs.
* `n_init`: How many times the K-Means algorithm is run internally with different random centroid seeds. It will automatically return the best model of the run.

# 10. Choosing the Number of Clusters

### Concept
Choosing $K$ randomly can result in a terrible model. Usually, the best $K$ is unknown. 

## The Elbow Method
We can measure how tightly clustered our groups are calculating the **Within-cluster sum of squares (Inertia)**.
We run K-Means multiple times with different $K$ values and plot $K$ vs Inertia. We look for an "elbow" or kink in the graph where improvements drastically slow down.


```python
# Generating data to test K values
X_elbow, _ = make_blobs(n_samples=500, centers=5, cluster_std=1.0, random_state=42)

inertias = []
K_range = range(1, 11)

for k in K_range:
    temp_model = KMeans(n_clusters=k, random_state=42, n_init=10)
    temp_model.fit(X_elbow)
    # Get inertia and append
    inertias.append(temp_model.inertia_)

# Plot the Elbow Method
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, marker='o', linestyle='--')
plt.title("The Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(K_range)
plt.axvline(x=5, color='r', linestyle=':', label='Elbow Point')
plt.legend()
plt.show()
```

### Interpretation
The plot sharply drops until it hits K=5. After K=5, adding more clusters only slightly reduces the Inertia. Therefore, 5 is our optimal K.


# 11. Silhouette Score

### Concept
* **What is Silhouette Score?** A metric calculating how well-separated clusters are from one another compared to how compact they are internally.
* **Scale**: Ranges from -1 to 1.
  * **Around +1**: Clusters are far apart and well-separated (Great).
  * **Around 0**: Overlapping clusters (Ambiguous).
  * **Negative (-1)**: Potentially poor clustering, assignments may be totally wrong.


```python
from sklearn.metrics import silhouette_score

silhouette_scores = []
# You need at least K=2 to calculate Silhouette Score
K_range_sil = range(2, 11)

for k in K_range_sil:
    temp_model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = temp_model.fit_predict(X_elbow)
    # Calculate score
    score = silhouette_score(X_elbow, labels)
    silhouette_scores.append(score)

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(K_range_sil, silhouette_scores, marker='s', linestyle='-', color='g')
plt.title("Silhouette Score vs K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(K_range_sil)
plt.show()
```

# 12. Choosing K with Elbow + Silhouette

### Concept
* **Elbow Method**: Shows you where adding clusters yields diminishing returns. Can sometimes be hard to read.
* **Silhouette Score**: Gives you an absolute validation metric of separation. It naturally peaks at the perfect mathematically separated clusters.

> Best Practice: Use **both** metrics simultaneously. They act as counter-balances. In our dataset, the Elbow point heavily indicated K=5, and the Silhouette Score absolutely definitively peaked at K=5. We confidently conclude K=5.


# 13. Feature Scaling and K-Means

### Concept
Distance calculations treat numeric values equivalently.
If you have `Age` varying from 20 to 60, and `Salary` varying from 20,000 to 100,000, calculating distance between them will be ruined because `Salary` will dominate the math. *Differences in Age will appear mathematically non-existent compared to Differences in Salary.*

> Important: You MUST scale data for K-Means using StandardScaler so every feature has a mean of 0 and std of 1.


```python
from sklearn.preprocessing import StandardScaler

# Unscaled Data: Huge differences in scale
feature1 = np.random.uniform(0, 10, 100)
feature2 = np.random.uniform(0, 10_000, 100)
X_unscaled = np.column_stack([feature1, feature2])

model_unscaled = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_unscaled)

# Scaled Data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_unscaled)
model_scaled = KMeans(n_clusters=3, random_state=42, n_init=10).fit(X_scaled)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# The unscaled clusters strictly group by the massive Y-axis ignoring the X-axis
ax1.scatter(X_unscaled[:,0], X_unscaled[:,1], c=model_unscaled.labels_, cmap='viridis')
ax1.set_title("Without Scaling (Dominated by Feature 2)")
ax1.set_xlabel("Feature 1 [0-10]")
ax1.set_ylabel("Feature 2 [0-10000]")

# The scaled features allows the algorithm to view both features evenly
ax2.scatter(X_scaled[:,0], X_scaled[:,1], c=model_scaled.labels_, cmap='viridis')
ax2.set_title("With StandardScaler (Balanced)")
ax2.set_xlabel("Scaled Feature 1")
ax2.set_ylabel("Scaled Feature 2")
plt.show()
```

# 14. K-Means Initialization

### Concept
* **Why centroid initialization matters**: Basic random initialization can drop two centroids right next to each other, resulting in a suboptimal local minimum (poor clusters).
* **Random initialization**: Dangerous due to blind luck.
* **K-Means++**: An intelligent default where centroids are initiated as far away from each other as possible. K-Means++ usually provides **better starting centroids** resulting in robust, reliable convergence.


```python
# Using K-Means++ explicitly (this is the default in Scikit-Learn anyway)
model_kpp = KMeans(n_clusters=3, init="k-means++", random_state=42, n_init=10)
print("Model initialized with K-Means++")
```

# 15. Random State and Reproducibility

### Concept
Because the first centroids are initialized with elements of randomness, running K-Means twice on identical code can sometimes yield slightly different setups (especially on unclear datasets).

By using `random_state=42`, we force the pseudorandom number generator into the same seed, causing our initial centroids to **always be identically generated on every rerun.** It ensures perfect reproducibility.


```python
# Reproducibility in action
model_reproducible = KMeans(n_clusters=3, random_state=42, n_init=10)
```

# 16. K-Means Limitations

### Concept
K-Means struggles heavily with certain configurations of data.

* **Need to choose K beforehand**: Requires iterating values manually.
* **Sensitive to initialization**: Requires K-means++ to protect against local minimums.
* **Sensitive to outliers**: Outliers corrupt centroids.
* **Sensitive to feature scaling**: Unscaled features ruin distance calculation.
* **Requires roughly spherical/compact clusters**: Struggles with elongated clouds, concentric rings, or moons.
* **Differing densities**: Struggles if one cluster is super dense and another is highly diffused.


```python
from sklearn.datasets import make_moons

X_moon, y_moon = make_moons(n_samples=200, noise=0.05, random_state=42)
km_moon = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X_moon)

plt.figure(figsize=(6, 4))
plt.scatter(X_moon[:, 0], X_moon[:, 1], c=km_moon.labels_, cmap='coolwarm')
plt.title("Limitation: K-Means cuts the moons awkwardly rather than curving")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

# 17. Outliers and K-Means

### Concept
Since the centroid is mathematically the *mean* of its assigned points, a massive outlier will wildly shift the position of that mean.


```python
X_normal = np.random.randn(20, 2)
# Add an extreme outlier far off
X_outlier = np.vstack([X_normal, [15, 15]])

km_normal = KMeans(n_clusters=1, random_state=42, n_init=10).fit(X_normal)
km_outlier = KMeans(n_clusters=1, random_state=42, n_init=10).fit(X_outlier)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.scatter(X_normal[:,0], X_normal[:,1], c='blue', label='Data')
ax1.scatter(km_normal.cluster_centers_[:,0], km_normal.cluster_centers_[:,1], c='red', s=200, marker='X', label='Centroid')
ax1.set_title("Without Outlier")
ax1.legend()

ax2.scatter(X_outlier[:,0], X_outlier[:,1], c='gray', label='Data (with outlier)')
ax2.scatter(km_outlier.cluster_centers_[:,0], km_outlier.cluster_centers_[:,1], c='red', s=200, marker='X', label='Shifted Centroid')
ax2.set_title("With Outlier")
ax2.legend()
plt.show()
```

# 18. K-Means vs Classification

### Concept
It's vital to remember that unsupervised clustering is entirely different from supervised classification (like Decision Trees or Logistic Regression).

| Target | K-Means | Classification |
| :--- | :--- | :--- |
| **Labels** | Labels are **not known**. Data is unlabeled. | Labels are **known** and fed to the model. |
| **Goal** | Group similar items together organically. | Predict distinct classes based on past data. |
| **Mapping** | Outputs arbitrary Integer IDs (0, 1, 2...). | Outputs semantic classes (Dog, Cat). |

> Common Mistake: When K-Means assigns a cluster label `0, 1, 2`, these numbers do **NOT automatically have semantic meaning**. The data scientist has to profile the group and assign meaning manually.


# 19. K-Means for Image Compression

### Concept
Images are fundamentally numerical grids. A color pixel contains three values (Red, Green, Blue) ranging from 0-255.
By reducing the amount of unique colors in an image (Color Quantization), we compress the image. K-Means performs this easily.

**Flow:**
Pixels → RGB values → Cluster similar colors into K groups → Replace pixels with their respective centroid colors.


```python
# 19. Simple Image Color Quantization Demo
# Generate a synthetic 50x50 "image" with smooth color gradients
y_img, x_img = np.mgrid[0:1:50j, 0:1:50j]
synthetic_image = np.zeros((50, 50, 3))
synthetic_image[:, :, 0] = x_img  # Red gradient
synthetic_image[:, :, 1] = y_img  # Green gradient
synthetic_image[:, :, 2] = 0.5    # Constant blue

# Reshape image to a list of pixels (2500 pixels, 3 color channels)
pixels = synthetic_image.reshape(-1, 3)

# Fit K-Means to find the 4 most representative colors
kmeans_img = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans_img.fit(pixels)

# Replace each pixel with its closest centroid color
compressed_pixels = kmeans_img.cluster_centers_[kmeans_img.labels_]
compressed_image = compressed_pixels.reshape(50, 50, 3)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(synthetic_image)
ax1.set_title("Original (Smooth Gradient)")
ax1.axis('off')

ax2.imshow(compressed_image)
ax2.set_title("Compressed (K=4 Colors)")
ax2.axis('off')
plt.show()
```

# 20. Complete End-to-End K-Means Project

### Goal: Segmenting Customers based on Annual Income and Spending Score.


```python
# 1. Create dataset
X_retail, _ = make_blobs(n_samples=500, centers=4, cluster_std=1.5, random_state=42)

# Adjust values to look like real business data
Income = np.abs(X_retail[:, 0] * 12 + 60) # Annual Income (k)
Spending = np.abs(X_retail[:, 1] * 10 + 50) # Spending Score (1-100)

# 2. Convert to DataFrame
df_customers = pd.DataFrame({'Annual Income': Income, 'Spending Score': Spending})

# 3. Inspect Data
display(df_customers.head())

# 4. Visualize Data (pre-clustering)
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_customers, x='Annual Income', y='Spending Score', color='gray')
plt.title("Customer Data - Raw")
plt.show()

# 5. Scale Features
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_customers)

# 6 & 7. Test K values & Elbow Method
inertias = []
for k in range(1, 10):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(df_scaled)
    inertias.append(km.inertia_)
    
# 8. Silhouette Score checking (briefly)
sil_scores = [silhouette_score(df_scaled, KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(df_scaled)) for k in range(2, 9)]

# Visualizing Evaluation
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(range(1, 10), inertias, marker='o')
ax1.set_title("Elbow Method")
ax1.set_xlabel("K")

ax2.plot(range(2, 9), sil_scores, marker='s', color='green')
ax2.set_title("Silhouette Score")
ax2.set_xlabel("K")
plt.show()

# 9. Select best K (Looking at Elbow (k=4) and Silhouette Peak (k=4))
optimal_k = 4

# 10. Train final K-Means
final_model = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
final_model.fit(df_scaled)

# 11. Assign cluster labels
df_customers['Cluster'] = final_model.labels_

# Un-scale centroids for interpretable visualization natively on the raw data scale
final_centroids_unscaled = scaler.inverse_transform(final_model.cluster_centers_)

# 12 & 13. Visualize clusters and plot centroids
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_customers, x='Annual Income', y='Spending Score', hue='Cluster', palette='Set1', s=60, alpha=0.8)
plt.scatter(final_centroids_unscaled[:, 0], final_centroids_unscaled[:, 1], color='black', marker='X', s=300, edgecolor='white', linewidth=2, label='Centroids')
plt.title("Final Customer Segments")
plt.legend()
plt.show()
```

# 21. Cluster Interpretation

### Finding Business Meaning
The model did the math, but now we must assign business meaning based on the features.
**Do not pretend cluster generated names are ground truth. Interpretations map feature behavior to real-world implications.**


```python
# 14. Describe each cluster
cluster_stats = df_customers.groupby('Cluster').agg(
    Number_of_Samples=('Annual Income', 'count'),
    Mean_Income=('Annual Income', 'mean'),
    Mean_Spending_Score=('Spending Score', 'mean')
).round(2)

display(cluster_stats)

# 15. Explain business interpretation (Programmatic generation based on the dataframe)
print("\n--- Business Interpretation ---")
for i in range(optimal_k):
    income = cluster_stats.loc[i, 'Mean_Income']
    spending = cluster_stats.loc[i, 'Mean_Spending_Score']
    
    inc_label = "High Income" if income > df_customers['Annual Income'].mean() else "Low Income"
    spend_label = "High Spending" if spending > df_customers['Spending Score'].mean() else "Low Spending"
    
    print(f"Cluster {i}: {inc_label} + {spend_label}")
```

# 22. Common K-Means Mistakes

### What can go wrong?

1. **Choosing K randomly**
   * *Problem:* Poor grouping.
   * *Better approach:* Use Elbow Method and Silhouette Score.
2. **Forgetting feature scaling**
   * *Problem:* Features with huge scales dominate the distance calculation.
   * *Better approach:* Always apply `StandardScaler`.
3. **Ignoring outliers**
   * *Problem:* Centroids are pulled away from the main group.
   * *Better approach:* Remove or cap extreme outliers before clustering.
4. **Assuming clusters are always meaningful**
   * *Problem:* K-means will *always* find clusters, even in totally random noise.
   * *Better approach:* Validate if the clusters make logical sense.
5. **Treating cluster numbers as labels**
   * *Problem:* Assuming Cluster 0 is "the best" just because it's 0.
   * *Better approach:* Profile the clusters by their means to label them.
6. **Running K-Means on inappropriate data**
   * *Problem:* Trying to use it on heavy categorical data or moon-shapes.
   * *Better approach:* Use specialized categorical clustering (like K-Modes) or density based (DBSCAN).
7. **Ignoring silhouette score**
   * *Problem:* Relying only on the elbow graph, which can be ambiguous.
   * *Better approach:* Cross-examine with silhouette score.
8. **Ignoring initialization**
   * *Problem:* Poor convergence due to random starts.
   * *Better approach:* Ensure `init='k-means++'` is used.
9. **Using too many features**
   * *Problem:* The "Curse of Dimensionality" causes distances to lose meaning.
   * *Better approach:* Use dimensionality reduction first if features > 50-100.
10. **Overinterpreting clusters**
    * *Problem:* Trying to force a complex business narrative on highly overlapping/weak clusters.
    * *Better approach:* Accept when data lacks strong inherent structure.


# 23. Interview Questions

1. **What is K-Means?** K-Means is an unsupervised machine learning algorithm that partitions data into a predefined number of distinct clusters based on feature similarity.
2. **What is clustering?** The process of grouping a set of objects such that objects in the same group are more similar to each other than to those in other groups.
3. **What does K represent?** It represents the number of clusters you want the algorithm to divide the dataset into.
4. **What is a centroid?** The calculated central geometric point (mean position) of a cluster.
5. **How does K-Means work?** It initializes centroids, assigns points to the nearest centroid, recalculates the centroids, and repeats until convergence.
6. **What distance metric is commonly used?** Euclidean Distance.
7. **What is inertia?** Also known as within-cluster sum of squares (WCSS), it measures the total squared distance of data points to their respective cluster centroids. Lower is tighter.
8. **What is the elbow method?** A heuristic where you plot $K$ vs Inertia. The "elbow" point, where adding clusters yields diminishing returns in inertia reduction, suggests the optimal $K$.
9. **What is silhouette score?** A metric from -1 to 1 representing how tightly points are grouped in their own cluster compared to neighboring clusters. Closer to 1 is better.
10. **Why is scaling important?** Because Euclidean distance is highly sensitive to features with large numerical ranges. Scaling ensures features contribute equally to the distance.
11. **What is K-Means++?** A smart initialization technique that spaces out initial centroids, avoiding poor local minima and improving convergence speed.
12. **Why can K-Means give different results?** K-Means relies heavily on the random initialization of the first centroids. Different starting points can yield different final clusters.
13. **What is `n_init`?** A parameter specifying how many times the K-Means algorithm runs with different random seed initializations. The best result is kept.
14. **What are the limitations?** Must choose K manually, sensitive to outliers, sensitive to scaling, assumes roughly spherical clusters.
15. **How does K-Means handle outliers?** It doesn't handle them well; outliers drag the centroid (the mean) toward themselves, misrepresenting the main group.
16. **Can K-Means determine K automatically?** No. K must be predefined by the user.
17. **K-Means vs classification?** Classification operates on labeled data to map inputs to known targets. K-Means uses unlabeled data to discover natural groupings.
18. **What happens if K is too small?** Distinct internal subgroups are violently forced together into a single broad cluster.
19. **What happens if K is too large?** A natural, coherent group is artificially splintered into multiple small fragments.
20. **When should K-Means not be used?** For distinct non-spherical shapes (e.g. spirals, moons), deeply overlapping density clouds, or datasets heavily comprised of categorical strings.
21. **Are categorical variables supported?** Standard K-Means expects numerical data to measure distance. Categorical data breaks the mathematical distance logic unless explicitly transformed (one-hot, etc.).
22. **What does convergence mean here?** It means the centroids have stopped moving between iterations; assignments stabilize.
23. **Is K-Means deterministic?** No, due to random centroid initialization. Setting a `random_state` allows deterministic outcomes.
24. **How bad is the Curse of Dimensionality for K-Means?** Very bad. In high dimensional space, Euclidean distance loses meaning because all points seem roughly equidistant. 
25. **What is the difference between Inertia and Silhouette score?** Inertia looks only at compactness (how tight a cluster is). Silhouette looks at compactness *and* separation (how far groups are from each other).
